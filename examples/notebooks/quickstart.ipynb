# silverpoint in a notebook

Each chart is a sandboxed frame: it draws in Jupyter, JupyterLab, VS Code and Databricks without trusting the notebook.

```python
import pandas as pd

import silverpoint

df = pd.DataFrame({"hour": ["00", "04", "08", "12", "16", "20"], "hits": [18, 11, 34, 71, 74, 62]})
silverpoint.line_chart(df, x="hour", value="hits", title="Hits per hour", curve="monotone")
```

```python
# Databricks: displayHTML(silverpoint.bar_chart(df, x="hour", value="hits").to_html(render="client"))
silverpoint.heatmap_chart(
    pd.DataFrame({"day": ["Mon", "Mon", "Tue"], "slot": ["am", "pm", "am"], "load": [3, 5, 2]}),
    row="day",
    column="slot",
    value="load",
)
```
